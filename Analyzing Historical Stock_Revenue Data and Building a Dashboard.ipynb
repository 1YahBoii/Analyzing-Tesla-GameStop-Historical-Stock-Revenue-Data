# Tesla & GameStop: Share Price vs Revenue

I pulled the full share-price history for Tesla and GameStop with `yfinance`, scraped each company's quarterly revenue from web pages, cleaned the revenue figures, and plotted price against revenue to see whether the two move together.

*Final project of IBM's **Python Project for Data Science** course (Coursera), documented in my own words.*

**Skills:** `yfinance`, web scraping (`requests`, BeautifulSoup, `pd.read_html`), cleaning text-formatted numbers, matplotlib.

## Libraries

```python
import yfinance as yf
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
import warnings
# Hide FutureWarnings to keep the output readable
warnings.filterwarnings("ignore", category=FutureWarning)
```

## The plotting function

One function draws both charts for a company: share price on top, quarterly revenue underneath, on a shared date axis. Both are cut at mid-2021 so the two companies cover the same window.

```python
import matplotlib.pyplot as plt

def make_graph(stock_data, revenue_data, stock):
    stock_data_specific = stock_data[stock_data.Date <= '2021-06-14']
    revenue_data_specific = revenue_data[revenue_data.Date <= '2021-04-30']

    fig, axes = plt.subplots(2, 1, figsize=(12, 8), sharex=True)

    # Stock price
    axes[0].plot(pd.to_datetime(stock_data_specific.Date), stock_data_specific.Close.astype("float"), label="Share Price", color="blue")
    axes[0].set_ylabel("Price ($US)")
    axes[0].set_title(f"{stock} - Historical Share Price")

    # Revenue
    axes[1].plot(pd.to_datetime(revenue_data_specific.Date), revenue_data_specific.Revenue.astype("float"), label="Revenue", color="green")
    axes[1].set_ylabel("Revenue ($US Millions)")
    axes[1].set_xlabel("Date")
    axes[1].set_title(f"{stock} - Historical Revenue")

    plt.tight_layout()
    plt.show()
```

## 1. Tesla share price

`yfinance` returns the complete daily price history as a DataFrame with the date as the index. I reset the index so `Date` becomes a normal column, which the plotting function needs.

```python
tesla = yf.Ticker('TSLA')
```

```python
tesla_data = pd.DataFrame(tesla.history(period=('max')))
```

```python
tesla_data.reset_index(inplace=True)
tesla_data.head()
```

## 2. Tesla revenue (web scraping)

Revenue isn't available through `yfinance`, so I scraped it from a web page that lists quarterly revenue. The page has several tables; `pd.read_html()` with `match="Tesla Quarterly Revenue"` picks out the right one.

```python
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0220EN-SkillsNetwork/labs/project/revenue.htm"
html_data = requests.get(url, timeout=10).text
```

```python
soup = BeautifulSoup(html_data, 'html.parser')
```

```python
from io import StringIO
tesla_revenue = pd.read_html(StringIO(html_data), match = "Tesla Quarterly Revenue")[0]
tesla_revenue.columns=["Date", "Revenue"]
```

The revenue arrives as text like `$1,234`, which can't be plotted. I stripped the `$` and commas, then dropped empty and missing rows.

```python
tesla_revenue["Revenue"] = tesla_revenue['Revenue'].str.replace(r"[,\$]", "", regex=True)
```

```python
tesla_revenue.dropna(inplace=True)

tesla_revenue = tesla_revenue[tesla_revenue['Revenue'] != ""]
```

```python
tesla_revenue.tail()
```

## 3. GameStop share price

Same approach as Tesla.

```python
GameStop = yf.Ticker('GME')
```

```python
gme_data = pd.DataFrame(GameStop.history(period='max'))
```

```python
gme_data.reset_index(inplace=True)
gme_data.head()
```

## 4. GameStop revenue (web scraping)

This page holds more than one quarterly-revenue table, so I matched on the table's CSS class as well as its title, then cleaned the column the same way.

```python
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0220EN-SkillsNetwork/labs/project/stock.html"
html_data_2 = requests.get(url, timeout=10).text
```

```python
soup = BeautifulSoup(html_data_2, 'html.parser')
```

```python
gme_revenue = pd.read_html(StringIO(html_data_2), attrs={"class": "historical_data_table table"}, match="Quarterly Revenue")[0]
gme_revenue.columns = ["Date", "Revenue"]
```

```python
gme_revenue["Revenue"] = gme_revenue["Revenue"].str.replace(r"[,\$]", "", regex=True)
gme_revenue = gme_revenue[gme_revenue["Revenue"] != ""]
gme_revenue.dropna(inplace=True)
```

```python
gme_revenue.tail()
```

## 5. The dashboard

```python
make_graph(tesla_data, tesla_revenue, 'Tesla')
```

**Tesla:** revenue climbs steadily from around 2013 and passes $10 billion a quarter by early 2021. The share price stays flat for years, then shoots up from 2020, rising far faster than revenue did.

```python
make_graph(gme_data, gme_revenue, 'GameStop')
```

**GameStop:** revenue is sharply seasonal (a spike every holiday quarter) and drifts downward after 2016. The share price, flat for nearly two decades, explodes in early 2021 while revenue was falling — the price move had nothing to do with the company's sales.

## What I took away

- Putting price and revenue on one shared time axis makes the comparison obvious in a way two separate tables never would.
- GameStop's 2021 spike is a clear case of share price disconnecting from business fundamentals.
- If a table already exists in the HTML, `pd.read_html()` with `match=` is far less code than looping with BeautifulSoup.
- Scraped numbers always arrive as text — cleaning them is part of the job, not an afterthought.
